```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("thyroid_dataset.csv")
```

```python
df.head()
```

```python
X = df.drop("Outlier_label", axis=1)
y = df["Outlier_label"]
```

```python
scaler = StandardScaler()
x_scaled = scaler.fit_transform(X)
```

```python
x_scaled
```

```python
from sklearn.ensemble import IsolationForest
```

```python
clf = IsolationForest(n_estimators=200, contamination=0.036, random_state=42)
```

```python
labels = clf.fit_predict(x_scaled)
```

```python
labels
```

```python
# Visualize
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X_pca = pca.fit_transform(x_scaled)
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(X_pca[:, 0], X_pca[:, 1], c=labels)
plt.xlabel("PC1")
plt.ylabel("PC2")
```

```python
import numpy as np

# outlier ditected
n_outliers = np.sum(labels == -1)
n_normals = np.sum(labels == 1)

print("outlier = ", n_outliers)
print("normal = ", n_normals)
```

### LOF

```python
from sklearn.neighbors import LocalOutlierFactor
```

```python
neighbors = LocalOutlierFactor(contamination=0.036)

labels = neighbors.fit_predict(x_scaled)
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(X_pca[:, 0], X_pca[:, 1], c=labels)
plt.xlabel("PC1")
plt.ylabel("PC2")
```

```python
# oulier ditected
n_outliers = np.sum(labels == -1)
n_normals = np.sum(labels == 1)

print("outlier = ", n_outliers)
print("normal = ", n_normals)
```
