```python
import seaborn as sns
from sklearn.datasets import make_blobs
```

```python
X, y = make_blobs(
    n_samples=1000,
    n_features=2,
    centers=4,
    random_state=42
)
```

```python
# Visualize the data
sns.scatterplot(x=X[:, 0], y=X[:, 1])
```

```python
# K-Means Clustering
from sklearn.cluster import KMeans
```

```python
K = 4

kmeans = KMeans(
    n_clusters=K,
    random_state=42
)
```

```python
kmeans
```

```python
#label = cluster number
labels = kmeans.fit_predict(X)
```

```python
sns.scatterplot(x=X[:, 0], y=X[:, 1], c=labels)
```

# Choose Our K value - Eblow, Silhouette score

## Eblow Method

```python
# Eblow Method

wcss = []
for k in range(1, 21):
    kmeans = KMeans(n_clusters=k)
    kmeans.fit_predict(X)
    wcss.append(kmeans.inertia_)
```

```python
sns.lineplot(x=range(1, 21), y=wcss, marker="o")  # wcss vs K plot
```

## Silhouette Score

```python
from sklearn.metrics import silhouette_score
```

```python
ss = []

for k in range(2, 21):
    kmeans = KMeans(n_clusters=k)
    labels = kmeans.fit_predict(X)
    score = silhouette_score(X, labels)

    ss.append(score)
```

```python
# plot - K & Silhouette score

sns.lineplot(x=range(2, 21), y = ss, marker = 'o')
```
